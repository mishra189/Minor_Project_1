```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from wordcloud import WordCloud

import re
```

```python
with open("hostel_bois.txt","r",encoding="utf-8") as file:
    chat=file.readlines()

print(chat[:10])
```

```python
pattern='(\d{1,2}/\d{1,2}/\d{2,4}),\s(\d{1,2}:\d{2}\s?[APMapm]{0,2})\s-\s'

messages=[]

for line in chat:
    if re.match(pattern,line):
        messages.append(line)

print(len(messages))
```

```python
df=pd.DataFrame(messages,columns=["Chat"])

df.head()
```

```python
df.drop_duplicates(inplace=True)

df.reset_index(drop=True,inplace=True)

df.head()
```

```python
print("Total Messages :",len(df))
```

```python
plt.figure(figsize=(10,5))

plt.plot(df.index)

plt.title("Message Timeline")

plt.show()
```
